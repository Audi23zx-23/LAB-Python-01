```python
saldo = float(50000)
deposito = float(input("Ingrese el valor a depositar: "))
retiro = float(input("Ingrese el valor del retiro: "))

saldo = saldo + deposito
saldo = saldo - retiro
interes = 0.02

saldo_intereses = saldo * interes
saldo_final = saldo_intereses + saldo

print("Saldo final con interes es: ", saldo_final)

# El estado inicial significa los valores y configuraciones necesarias para que
# el programa funcione de forma correcta, mientras que el estado final es el que
# se obtiene despues de las modificaciones del programa.
```
