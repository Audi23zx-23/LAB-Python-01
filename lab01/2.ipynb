```python
# Lista inicial de temperaturas en °C
temperaturas = [22.5, 23.0, 21.8, 24.3, 25.1, 23.7]

# Operaciones básicas
# Agregar temperatura al final
temperaturas.append(39.5)
print(temperaturas)
# Eliminar primera temperatura
del temperaturas[0]
print(temperaturas)
# Insertar nueva temeperatura en la posicion 2
temperaturas.insert(1, 23.3)
print(temperaturas)
# Ordenar la lista de menor a mayor
temperaturas.sort()
print(temperaturas)
# Calcular diferencia de extremos y promedio de todos los valores
maximo = max(temperaturas)
minimo = min(temperaturas)

diferencia = maximo - minimo
promedio = sum(temperaturas) / len(temperaturas)

print("Diferencia: ", diferencia)
print("Promedio: ", promedio)
# Encuentre la posicion del valor 23.7
print(temperaturas.index(23.7))

# Muestra las temeperaturas de la posicion 2 y 4
print(temperaturas[1])
print(temperaturas[3])

# Muestra las tres primeras temeperaturas

a, b, c = temperaturas[0], temperaturas[1], temperaturas[2]
print(a, b, c)
print("Temperaturas:", temperaturas)
```
