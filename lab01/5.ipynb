```python
biblioteca = {
    "libros": [
        {"titulo": "1984", "autor": "George Orwell", "prestado": False},
        {"titulo": "El Principito", "autor": "Antoine de Saint-Exupéry", "prestado": True},
    ],
    "usuarios": ["Ana", "Luis", "María"]
}

# Muestra cuantos libros estan prestados y cuantos disponibles
libros_prestados = 0
for libro in biblioteca["libros"]:
    if libro["prestado"]:
        libros_prestados += 1
libros_disponibles = len(biblioteca["libros"]) - libros_prestados
print(f"Hay {libros_prestados} libros prestados y {libros_disponibles} libros disponibles.")

# Agregar nuevo libro
biblioteca["libros"].append({"titulo": "Cien años de soledad", "autor": "Gabriel García Márquez", "prestado": False})

# Marcar como prestado el libro de 1984
for libro in biblioteca["libros"]:
    if libro["titulo"] == "1984":
        libro["prestado"] = True
        break

# Mostrarel estado actual del sistema
print("Estado actual del sistema:")
print(biblioteca)
```
