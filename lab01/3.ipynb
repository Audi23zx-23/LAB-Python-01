```python
# Lista de productos en una tienda
inventario = [
    {"nombre": "Pan", "precio": 1500, "stock": 30},
    {"nombre": "Leche", "precio": 3500, "stock": 15},
    {"nombre": "Café", "precio": 8000, "stock": 10},
]
# Operaciones
# Muestra el nombre del primer producto
print(inventario[0]["nombre"])

#Agregar nuevo Producto
inventario.append({'nombre': "Doritos",'precio': 5000, 'stock': 30})
print(inventario)

#disminuir stock de leche
inventario[1]["stock"] -= 2
print(inventario[1]["stock"])

# Calcular el valor total del inventario
valor_total = sum(producto["precio"] * producto["stock"] for producto in inventario)
print(f"El valor total del inventario es: {valor_total}")

# Mostrar productos cuyo stock sea mayor a 10
productos_stock_mayor_a_10 = [producto for producto in inventario if producto["stock"] > 10]
print("Productos con stock mayor a 10:")
for producto in productos_stock_mayor_a_10:
    print(f"{producto['nombre']} - Stock: {producto['stock']}")

# Aumentar precio un 10%
for producto in inventario:
    producto["precio"] *= 1.1
print("Precio aumentado en un 10%:")
for producto in inventario:
    print(f"{producto['nombre']} - Precio: {producto['precio']}")

# Redondeo de precios para evitar valores largos
for producto in inventario:
    producto["precio"] = round(producto["precio"])
print("Precios redondeados:")
for producto in inventario:
    print(f"{producto['nombre']} - Precio: {producto['precio']}")

# Buscar productos por nombre indicado por el usuario y mostrarlo (Tener en cuenta que puede no exisitir)
nombre_buscado = input("Ingrese el nombre del producto a buscar: ")
producto_encontrado = next((producto for producto in inventario if producto["nombre"] == nombre_buscado), None)
if producto_encontrado:
    print(f"Producto encontrado: {producto_encontrado['nombre']} - Precio: {producto_encontrado['precio']}")
else:
    print("Producto no encontrado.")

# Eliminar producto del inventario en la posicion 0
del inventario[0]
print("Inventario después de eliminar el primer producto:")
for producto in inventario:
    print(f"{producto['nombre']} - Precio: {producto['precio']}")

# Ordenar lista de forma ascendente
inventario.sort(key=lambda x: x['precio'])
print("Inventario ordenado por precio:")
for producto in inventario:
    print(f"{producto['nombre']} - Precio: {producto['precio']}")
```
