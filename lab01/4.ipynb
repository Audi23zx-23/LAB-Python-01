```python
# Tupla: coordenadas (x, y)
punto = (3, 7)
print("Punto:", punto)

# Crear una tupla rectangulo = (ancho, alto).
rectangulo = (4, 6)
# Calcular area y perimetro
area = rectangulo[0] * rectangulo[1]
perimetro = 2 * (rectangulo[0] + rectangulo[1])
print("Area:", area)
print("Perimetro:", perimetro)

# Calcula la distancia entre estos puntos
import math
posiciones = [(0, 0), (1, 2), (3, 4)]

print("Distancias entre puntos:")
for i in range(len(posiciones) - 1):
    punto1 = posiciones[i]
    punto2 = posiciones[i+1]

    distancia = math.sqrt(math.pow(punto2[0] - punto1[0], 2) + math.pow(punto2[1] - punto1[1], 2))
    print(f"Distancia entre {punto1} y {punto2}: {distancia:.2f}")
```
