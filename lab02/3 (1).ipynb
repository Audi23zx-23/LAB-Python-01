```python
print("Programa para sumar números. Ingresa 0 para terminar.")
suma_total = 0
numero = -1

while numero != 0:
    try:
        numero = int(input("Ingresa un número: "))
        suma_total += numero
    except ValueError:
        print("Entrada inválida. Por favor, ingresa un número entero.")

print(f"La suma total de los números ingresados es: {suma_total}")
```

En el programa que solicita números hasta que el usuario ingresa 0:

El estado del ciclo se representa principalmente por la variable numero. Esta variable se actualiza en cada iteración con la entrada del usuario.
