```python
productos = []
while True:
    nombre = input("Ingrese producto (o 'fin' para salir): ")
    if nombre == "fin":
        break

    try:
        precio = float(input("Precio: "))
        cantidad = int(input("Cantidad: "))
    except ValueError:
        print("Entrada inválida. Por favor, ingrese un número para precio y cantidad.")
        continue

    if precio < 0 or cantidad < 0:
        print("Error: El precio y la cantidad no pueden ser negativos. Intente de nuevo.")
        continue

    productos.append({"nombre": nombre, "precio": precio, "cantidad": cantidad})

# Mostrar total
total = 0
for p in productos:
    total += p["precio"] * p["cantidad"]

print("Total del inventario:", total)

# Mostrar el producto más caro
if productos:
    producto_mas_caro = max(productos, key=lambda p: p["precio"])
    print("\nProducto más caro:", producto_mas_caro)

# Representar el estado final del inventario como lista de diccionarios
print("\nEstado final del inventario:", productos)
```
