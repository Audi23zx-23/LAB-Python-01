```python
suma_pares = 0
suma_impares = 0

for i in range(1, 11):
    if i % 2 == 0:
        suma_pares += i
    else:
        suma_impares += i

# Representa el estado final como un diccionario
estado_final = {"suma_pares": suma_pares, "suma_impares": suma_impares}
print("Estado final:", estado_final)
```
